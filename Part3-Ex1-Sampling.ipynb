```python
import numpy as np
```

```python
import pandas as pd
```

```python
population = pd.DataFrame(np.arange(10)+1)
```

```python
population
```

```python
population.mean()
```

```python
population.std()
```

```python
# With random_state we fix the random generator (sampling without replacement)
s1 = population.sample(n=3, random_state=1)
s1
```

```python
s1.mean()
```

```python
s1.std()
```

```python
s2 = population.sample(n=3, random_state=2)
```

```python
s2
```

```python
s2.mean()
```

```python
s2.std()
```

```python
# With random_state we fix the random generator (sampling with replacement)
s3 = population.sample(frac=0.5, replace=True, random_state=1)
```

```python
s3
```

```python
s3.mean()
```

```python
s3.std()
```

```python
s4 = population.sample(frac=0.5, replace=True, random_state=2)
```

```python
s4
```

```python
s4.mean()
```

```python
s4.std()
```
